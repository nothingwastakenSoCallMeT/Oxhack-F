```python
from src.data import load

for t in ["SPY", "TLT", "GLD"]:
    df = load(t)
    r = df["Close"].pct_change().dropna()
    print(t,
          "dupes:", df.index.duplicated().sum(),
          "nans:", int(df.isna().sum().sum()),
          "min close:", round(df["Close"].min(), 2),
          "worst/best day:", f"{r.min():.1%} / {r.max():.1%}",
          "ann. vol:", f"{r.std() * 252**0.5:.1%}")
          
```
